- **Model (CSF):** geometry and material.
- **Problem:** loading and boundary conditions.
- **Case (CUF):** numerical approximations and output.

Run the cells in order. Then change a value in a YAML file and run the analysis again. The full explanation is in the [Quick Start documentation](https://github.com/giovanniboscu/continuous-section-field/blob/main/cuf/quickstart/README.md).

### 1 · Prepare the environment

This cell downloads the original repository and installs `csfpy` (which includes CSF-CUF) and `pypardiso`. Run it **once per new Colab session**.

```python
from pathlib import Path
import subprocess
import sys

REPO = Path('/content/continuous-section-field')
if not REPO.exists():
    subprocess.run(
        ['git', 'clone', '--depth', '1',
         'https://github.com/giovanniboscu/continuous-section-field.git', str(REPO)],
        check=True,
    )

subprocess.run(
    [sys.executable, '-m', 'pip', 'install', '-q', '-e', str(REPO), 'pypardiso'],
    check=True,
)

QUICKSTART = REPO / 'cuf' / 'quickstart'
assert QUICKSTART.is_dir(), f'Quick Start not found: {QUICKSTART}'
print('Ready. Quick Start:', QUICKSTART)
```

### 2 · Explore the original three YAML files

The files below are **the actual Quick Start files in the cloned repository**, not copies maintained by this notebook. This cell displays them; it does not modify them.

| File | What it defines |
|---|---|
| `models/rectangular_prismatic_csf.yaml` | Rectangular section, beam length, elastic material |
| `problems/torsion_halfwave.yaml` | Torsional load and the predefined problem |
| `cases/torsion_halfwave_legendre_N08.yaml` | Problem adapter, CUF expansion, longitudinal FEM, output |


```python
FILES = {
    'MODEL (CSF)': QUICKSTART / 'models' / 'rectangular_prismatic_csf.yaml',
    'PROBLEM': QUICKSTART / 'problems' / 'torsion_halfwave.yaml',
    'CASE (CUF)': QUICKSTART / 'cases' / 'torsion_halfwave_legendre_N08.yaml',
}

for label, path in FILES.items():
    print(f'\n--- {label}: {path.relative_to(QUICKSTART)} ---')
    print(path.read_text(encoding='utf-8'))
```

### 3 · Change one number (optional)

In Colab, click the **folder icon** on the left (**Files**). Open `continuous-section-field/cuf/quickstart/`, then open a YAML file in the text editor and **save** your change.

Try just one of these changes:

- **Material:** change `weight: 71700.0` in the **model** (MPa in this example).
- **Geometry:** change one `[y, z]` vertex coordinate in the **model** (mm). Change the matching vertex in both `S0` and `S1` if you want the section to remain prismatic.
- **CUF:** change `cuf: order: 8` to another order in the **case** (the `order` nested under `cuf`, not the longitudinal `order`).
- **Load:** change `amplitude: 10.0` in the **problem** (N/mm).

**Nothing else needs changing:** the solver reads the saved YAML files when you run the next cell. To inspect the updated text, rerun cell 2. Edits inside Colab's temporary runtime are not saved back to GitHub automatically.

### 4 · Run the Quick Start

Same command and same case as the repository Quick Start. After editing a YAML file, **rerun only this cell**.

```python
subprocess.run(
    ['csf-cuf', 'cases/torsion_halfwave_legendre_N08.yaml'],
    cwd=QUICKSTART,
    check=True,
)

RESULTS = QUICKSTART / 'output' / 'torsion_halfwave_legendre_N08'
print(f'\nResults directory: {RESULTS}')
if RESULTS.exists():
    result_files = sorted(p.relative_to(RESULTS) for p in RESULTS.rglob('*') if p.is_file())
    for path in result_files[:15]:
        print(' -', path)
    if len(result_files) > 15:
        print(f' ... and {len(result_files) - 15} more files')
else:
    print('No output directory found at the expected Quick Start location.')
```

**That is all.** Change geometry, material, load, or the CUF order, save the YAML, and run step 4 again.

If you want to **reset everything to the original Quick Start**, start a fresh Colab runtime and rerun step 1. For a theoretical explanation, follow the [main CSF-CUF documentation](https://github.com/giovanniboscu/continuous-section-field/blob/main/cuf/readme.md).
